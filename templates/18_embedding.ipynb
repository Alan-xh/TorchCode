# 🟢 简单：嵌入层

从零实现一个**嵌入查找表**。num_embeddings表示词表的大小，embedding_dim表示每个字的嵌入维度。

### 函数签名
```python
class MyEmbedding(nn.Module):
    def __init__(self, num_embeddings: int, embedding_dim: int): ...
    def forward(self, indices: Tensor) -> Tensor: ...
```

### 规则
- `self.weight`：形状为 `(num_embeddings, embedding_dim)` 的 `nn.Parameter`
- 前向传播：对权重矩阵进行索引 —— `weight[indices]`
- 禁止使用 `nn.Embedding`

````python
import torch
import torch.nn as nn
from torch import Tensor

class MyEmbedding(nn.Module):
    def __init__(self, num_embeddings: int, embedding_dim: int):
        super().__init__()
        # 初始化权重矩阵，形状为 (num_embeddings, embedding_dim)
        # 使用均匀分布初始化，与 PyTorch 的默认行为一致
        self.weight = nn.Parameter(
            torch.randn(num_embeddings, embedding_dim) * 0.1
        )
        # 也可以用更标准的初始化：
        # nn.init.uniform_(self.weight, -1.0, 1.0)
        
    def forward(self, indices: Tensor) -> Tensor:
        # 直接使用索引从权重矩阵中提取对应的嵌入向量
        # 支持任意形状的索引张量（包括多维）
        return self.weight[indices]
````

````python
# 🧪 调试
emb = MyEmbedding(10, 4)
idx = torch.tensor([0, 3, 7])
print('输出形状:', emb(idx).shape)
print('与手动索引一致:', torch.equal(emb(idx)[0], emb.weight[0]))
````
