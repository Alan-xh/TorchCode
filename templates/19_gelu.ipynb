`erf` 是**误差函数**（Error Function）的缩写，是一个在统计学和数学中非常重要的特殊函数。

$$\text{erf}(x) = \frac{2}{\sqrt{\pi}} \int_0^x e^{-t^2} dt$$

- **值域**：(-1, 1)
- **奇函数**：erf(-x) = -erf(x)
- **极限**：
  - 当 x → ∞ 时，erf(x) → 1
  - 当 x → -∞ 时，erf(x) → -1
- **特殊值**：erf(0) = 0

误差函数与标准正态分布的**累积分布函数**（CDF）Φ(x) 密切相关：

$$\Phi(x) = \frac{1}{2}\left(1 + \text{erf}\left(\frac{x}{\sqrt{2}}\right)\right)$$

GELU 使用 erf 来近似标准正态分布的累积分布函数：

$$\text{GELU}(x) = x \cdot \Phi(x) = x \cdot 0.5 \cdot (1 + \text{erf}(x / \sqrt{2}))$$

这样使得 GELU 在 x=0 附近近似线性，在两端则像 ReLU 但有平滑的过渡。

PyTorch 的 `torch.erf` 使用高效且精确的数值算法，保证了：
- 高精度计算
- 自动微分支持
- GPU 加速

````python
import torch
from torch import Tensor

def my_gelu(x: Tensor) -> Tensor:
    """
    GELU (Gaussian Error Linear Unit) 激活函数
    
    公式: GELU(x) = x * Φ(x) = x * 0.5 * (1 + erf(x / sqrt(2)))
    
    Args:
        x: 输入张量
        
    Returns:
        应用 GELU 激活后的张量
    """
    # 计算 x / sqrt(2)
    x_scaled = x / torch.sqrt(torch.tensor(2.0))
    
    # 使用 torch.erf 计算误差函数
    erf_result = torch.erf(x_scaled)
    
    # 计算 Φ(x) = 0.5 * (1 + erf(x / sqrt(2)))
    phi = 0.5 * (1 + erf_result)
    
    # GELU(x) = x * Φ(x)
    return x * phi
````

````python
# 🧪 调试
x = torch.tensor([-2., -1., 0., 1., 2.])
print('输出:', my_gelu(x))
print('参考:', torch.nn.functional.gelu(x))
````
