# 🟠 中等：带预热的余弦学习率调度器

实现一个 **余弦学习率调度**，带线性预热。

### 函数签名
```python
def cosine_lr_schedule(step, total_steps, warmup_steps, max_lr, min_lr=0.0) -> float:
```

### 调度
```
step < warmup:  lr = max_lr * step / warmup_steps  (线性递增)
step >= warmup: lr = min_lr + 0.5*(max_lr-min_lr)*(1 + cos(π * progress))
```
其中 `progress = (step - warmup) / (total - warmup)`

````python
def cosine_lr_schedule(step, total_steps, warmup_steps, max_lr, min_lr=0.0) -> float:
    """
    带线性预热的余弦学习率调度器
    
    Args:
        step: 当前训练步数（从0开始）
        total_steps: 总训练步数
        warmup_steps: 预热步数
        max_lr: 最大学习率
        min_lr: 最小学习率（默认0.0）
    
    Returns:
        float: 当前步的学习率
    """
    # 边界情况处理
    if total_steps <= 0 or warmup_steps < 0:
        return min_lr
    
    # 如果步数小于0，返回最小学习率
    if step < 0:
        return min_lr
    
    # 线性预热阶段
    if step < warmup_steps:
        # 线性递增：从0到max_lr
        return max_lr * step / warmup_steps if warmup_steps > 0 else max_lr
    
    # 余弦退火阶段
    # 计算在余弦阶段的进度 (0到1)
    progress = (step - warmup_steps) / (total_steps - warmup_steps)
    # 限制progress在[0, 1]范围内
    progress = min(progress, 1.0)
    
    # 余弦退火公式
    lr = min_lr + 0.5 * (max_lr - min_lr) * (1 + math.cos(math.pi * progress))
    
    return lr
````

````python
# 🧪 调试
lrs = [cosine_lr_schedule(i, 100, 10, 0.001) for i in range(101)]
print(f'起始: {lrs[0]:.6f}')
print(f'预热结束: {lrs[10]:.6f}')
print(f'中间: {lrs[55]:.6f}')
print(f'结束: {lrs[100]:.6f}')
````
