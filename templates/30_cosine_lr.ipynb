# 🟠 中等：带预热的余弦学习率调度器

实现一个 **余弦学习率调度**，带线性预热。

### 函数签名
```python
def cosine_lr_schedule(step, total_steps, warmup_steps, max_lr, min_lr=0.0) -> float:
```

### 调度
```
step < warmup:  lr = max_lr * step / warmup_steps  (线性递增)
step >= warmup: lr = min_lr + 0.5*(max_lr-min_lr)*(1 + cos(π * progress))
```
其中 `progress = (step - warmup) / (total - warmup)`

**阶段一：线性预热（Linear Warmup）min_lr -> max_lr**

防止梯度爆炸：在训练初期，模型权重是随机初始化的。此时如果直接用较大的学习率，梯度会非常大，导致损失剧烈震荡甚至发散（梯度爆炸）。\
稳定梯度统计：在使用 Batch Normalization（批归一化） 或 Layer Normalization（层归一化） 时，前几步的均值和方差统计量不稳定。小学习率可以让这些统计量平稳地适应数据。\
慢慢进入状态：这好比汽车启动时，不是直接猛踩油门，而是缓慢加速，让各个部件（权重）适应。

lr = base_lr * (当前步数 / 预热总步数)  

**阶段二：余弦退火（Cosine Annealing） 0.5 * (max_lr - min_lr) * 2 -> 0.5 * (max_lr - min_lr)**

精细调优：在训练后期，模型接近最优点（Loss 最低点）。如果学习率还是很大，就会在最优解附近反复“震荡”而无法真正抵达。我们需要让学习率慢慢降下来，像走下山坡一样小心翼翼地靠近谷底。\
平滑下降：相比 Step Decay（阶梯式下降，比如每 30 个 epoch 除以 10），余弦退火的下降曲线更加平滑，不会让学习率突然“跳崖”，训练过程更稳定。

lr = 0.5 * base_lr * (1 + cos( π * (当前步数 - 预热步数) / (总步数 - 预热步数) ))  

````python
def cosine_lr_schedule(step, total_steps, warmup_steps, max_lr, min_lr=0.0) -> float:
    """
    带线性预热的余弦学习率调度器
    
    Args:
        step: 当前训练步数（从0开始）
        total_steps: 总训练步数
        warmup_steps: 预热步数
        max_lr: 最大学习率
        min_lr: 最小学习率（默认0.0）
    
    Returns:
        float: 当前步的学习率
    """
    # 边界情况处理
    if total_steps <= 0 or warmup_steps < 0:
        return min_lr
    
    # 如果步数小于0，返回最小学习率
    if step < 0:
        return min_lr
    
    # 线性预热阶段
    if step < warmup_steps:
        # 线性递增：从 0 到 max_lr
        return max_lr * step / warmup_steps if warmup_steps > 0 else max_lr
    
    # 余弦退火阶段
    # 计算在余弦阶段的进度 (0到1)
    progress = (step - warmup_steps) / (total_steps - warmup_steps)
    # 限制progress在[0, 1]范围内
    progress = min(progress, 1.0)
    
    # 余弦退火公式
    lr = min_lr + 0.5 * (max_lr - min_lr) * (1 + math.cos(math.pi * progress))
    
    return lr
````

````python
# 🧪 调试
lrs = [cosine_lr_schedule(i, 100, 10, 0.001) for i in range(101)]
print(f'起始: {lrs[0]:.6f}')
print(f'预热结束: {lrs[10]:.6f}')
print(f'中间: {lrs[55]:.6f}')
print(f'结束: {lrs[100]:.6f}')
````
