```python
import torch
```

**tensor.permute()** 置换 是 PyTorch 中用于交换张量维度顺序的核心函数。它不会改变张量的数据本身，只会改变数据的视图（view）或返回一个重新排列步长（stride）的副本。permute 返回的结果通常与原张量共享底层数据内存（即浅拷贝）

```python
# 创建一个形状为 (2, 3, 4) 的三维张量 (批次, 高度, 宽度)
x = torch.randn(2, 3, 4)

# 将维度顺序改为 (宽度, 高度, 批次) -> 形状变为 (4, 3, 2)
y = x.permute(2, 1, 0) 
print(y.shape)  # 输出: torch.Size([4, 3, 2])
```
